```python
import os
from pathlib import Path

import geopandas as gpd

from amazonas_pipeline.utils import join_cells_with_polygons
```

```python
data_path = Path(os.environ["DATA_PATH"])

generated_path = data_path / "generated"

ghsl_path = Path(os.environ["GHSL_PATH"])

cells_path = Path("./generated/cells")
cells_path.mkdir(exist_ok=True, parents=True)

df_final_path = Path("./generated/df_final")
```

```python
for year in range(1975, 2021, 5):
    df_final = gpd.read_file(df_final_path / f"{year}.gpkg")

    df_cells = add_pop_and_smod_to_cells(
        join_cells_with_polygons(
            gpd.read_file(generated_path / "cells" / "countries.gpkg"),
            df_final,
        ),
        selected_year=year,
    )
    df_cells.to_file(cells_path / f"{year}.gpkg", driver="GPKG")
```
