```python
import os
from pathlib import Path

import pandas as pd
```

```python
sent_path = Path(os.environ["SENT_PATH"])
year_path = sent_path / "DEGURBA" / "hojas" / "1975"
```

```python
df = pd.read_excel(year_path / "normal.xlsx")

temp = df.loc[lambda df: df["pop_urban_cluster_1975"] > 0]
temp_in_amazon = temp.loc[lambda df: df["in_amazon"] == "yes"].assign(
    pop_1975_edit=lambda df: (
        df["pop_urban_cluster_1975"]
        + df["pop_urban_center_1975"]
        + df["pop_rural_1975"]
    )
)
temp_out_amazon = temp.loc[lambda df: df["in_amazon"] == "no"].assign(
    pop_1975_edit=lambda df: df["pop_urban_cluster_1975"] + df["pop_urban_center_1975"]
)

temp_normal = pd.concat([temp_in_amazon, temp_out_amazon]).loc[
    lambda df: df["pop_1975_edit"] > 5000
]
```

```python
df = pd.read_excel(year_path / "sin_conurb.xlsx")

temp = df.loc[lambda df: df["pop_urban_cluster_1975"] > 0]
temp_in_amazon = temp.loc[lambda df: df["in_amazon"]].assign(
    pop_1975_edit=lambda df: (
        df["pop_urban_cluster_1975"]
        + df["pop_urban_center_1975"]
        + df["pop_rural_1975"]
    )
)
temp_out_amazon = temp.loc[lambda df: ~df["in_amazon"]].assign(
    pop_1975_edit=lambda df: df["pop_urban_cluster_1975"] + df["pop_urban_center_1975"]
)

temp_sin_conurb = pd.concat([temp_in_amazon, temp_out_amazon]).loc[
    lambda df: df["pop_1975_edit"] > 5000
]
```
